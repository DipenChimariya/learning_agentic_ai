```python

from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Initialize the Gemini Model
#the super fast and stable gemini-2.5-flash
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

response = llm.invoke([
    ["system","you're a helpful assistant that answer in short 2-3 lines"],
    ["human","provide a motivational line for a Agentic AI learner like me"]
])
print(response.text)
```

```python
prompt=([
    ["system","you're a professional AI engineer, who answers in short 2-3 lines"],
    ["human","what is an agentic ai?"]
])

llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0)

response=llm.invoke(prompt)

print(response.text)
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=1)
response=llm.invoke("Say a short 1-2 lines funny joke")
print(response.text)
```

```python
#calling llm using groq model
from langchain_groq import ChatGroq
llm=ChatGroq(model="qwen/qwen3-32b",reasoning_format="parsed")

response=llm.invoke([
    ["system", "you're a helpful assistant who provide short 2-3 lines answer"],
    ["human","what is groq ai?"]
])

print(response.content)
```
