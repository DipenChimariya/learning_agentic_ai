```python
import chromadb

# 1. Start the ephemeral memory client
client = chromadb.Client()

# 2. Spin up your vector collection sandbox
collection = client.create_collection("fact_articles")

# 3. Add strings to the engine
collection.add(
    ids=["id1", "id2", "id3", "id4", "id5"],
    documents=[
        "AI is a future in tech",
        "Learning new skill is an art",
        "Spacex is a really good spacecraft company",
        "There is very good scope if anyone knows trading",
        "Money is just a tool to live a good life"
    ]
)

# FIX: Explicitly request 'embeddings' in the include list parameter
data = collection.get(
    include=["documents", "embeddings"]
)

# 4. Safely loop and inspect the internal representations
for i, emb in enumerate(data["embeddings"]):
    print(f"\n📄 Document {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding length: ", len(emb))
    print("First 10 vector values: ", emb[:10])
```
